```python
# LEt us do the model buildiung using K-means Algorithm without standardisation

from sklearn.cluster import KMeans

# Step 2: Initialize and fit the model
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(df[['SepalLengthCm', 'SepalWidthCm','PetalLengthCm','PetalWidthCm','Species_label_encoded']])  # Model is trained here
```

```python
# Step 3: Predict clusters
df['KMeans_cluster'] = kmeans.predict(df[['SepalLengthCm', 'SepalWidthCm','PetalLengthCm','PetalWidthCm','Species_label_encoded']])
```
