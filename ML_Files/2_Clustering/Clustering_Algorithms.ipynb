```python
1. Clustering Algorithms
Used to group similar data points together
```

```python
| Algorithm                         | Description                                                                             | Use Case                 |
| --------------------------------- | --------------------------------------------------------------------------------------- | ------------------------ |
| **K-Means**                       | Partitions data into K clusters based on distance.                                      | Customer segmentation    |
| **Hierarchical Clustering**       | Builds a tree (dendrogram) of clusters.                                                 | Gene expression analysis |
| **DBSCAN**                        | Density-based clustering that can find arbitrarily shaped clusters and detect outliers. | Spatial data clustering  |
| **Mean Shift**                    | Moves data points to the mode of the data distribution.                                 | Image segmentation       |
| **Gaussian Mixture Models (GMM)** | Soft clustering using a mixture of Gaussians.                                           | Speaker recognition      |
| **Affinity Propagation**          | Message passing between data points to find exemplars.                                  | Image classification     |
```

```python
2. Dimensionality Reduction Algorithms
Used to reduce the number of input variables/features.
```

```python
| Algorithm                                                | Description                                                  | Use Case                             |
| -------------------------------------------------------- | ------------------------------------------------------------ | ------------------------------------ |
| **PCA (Principal Component Analysis)**                   | Projects data onto lower dimensions to maximize variance.    | Data visualization                   |
| **t-SNE (t-Distributed Stochastic Neighbor Embedding)**  | Non-linear dimensionality reduction for visualization.       | High-dimensional image data          |
| **UMAP (Uniform Manifold Approximation and Projection)** | Preserves local and global structure better than t-SNE.      | Embedding visualization              |
| **Autoencoders**                                         | Neural networks that learn to compress and reconstruct data. | Anomaly detection, image compression |
| **ICA (Independent Component Analysis)**                 | Separates a multivariate signal into additive subcomponents. | Signal processing                    |
```

```python
3. Association Rule Learning
Discovers interesting relationships (associations) between variables
```

```python
| Algorithm   | Description                                              | Use Case                |
| ----------- | -------------------------------------------------------- | ----------------------- |
| **Apriori** | Generates frequent itemsets and association rules.       | Market basket analysis  |
| **Eclat**   | Uses depth-first search and tidsets for rule generation. | Retail transaction data |
```

```python
4. Neural Network-Based Unsupervised Models
Modern deep learning unsupervised approaches
```

```python
| Algorithm                                  | Description                                         | Use Case                            |
| ------------------------------------------ | --------------------------------------------------- | ----------------------------------- |
| **Autoencoders**                           | Learn efficient representations.                    | Image denoising, anomaly detection  |
| **Variational Autoencoders (VAE)**         | Probabilistic version of autoencoders.              | Generative modeling                 |
| **Generative Adversarial Networks (GANs)** | Two networks (generator and discriminator) compete. | Image generation, data augmentation |
```
